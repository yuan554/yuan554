```python
source = open("C:\\Users\\yuan\\Desktop\\lecture 6\\jingdong.txt","r",encoding="gbk")
line1 = source.readlines()
line1
```

```python
sentimentslist = []
for i in line1:
    s = SnowNLP(i)
    print(s.sentiments)
    sentimentslist.append(s.sentiments)
```

```python
#绘制情感分析条形图
import matplotlib.pyplot as plt
import pandas
import numpy
plt.hist(sentimentslist, bins = np.arange(0, 1, 0.01), facecolor = 'y')#bins参数为数据拆分间隔
plt.xlabel('Sentiments Probability')#添加x轴坐标名
plt.ylabel('Quantity')#添加y轴坐标名
plt.title('Analysis of Sentiments')#添加标题
plt.show()#显示图片
```

```python
df = pd.read_csv("D:/jingdong.csv",encoding="gbk")
df["sentiment"] =sentimentslist #指定刚情感得分作为原始表新的一列以便进行dataframe相关操作
df
```

```python
df1=df.drop("Unnamed: 1",axis=1)
df1
```

```python
df1.sentiment.describe()#描述性统计分析
```

```python
#绘制情感分析瀑布图
y=df["sentiment"]
plt.figure(figsize=(20, 8), dpi=800)
plt.plot(y)
plt.show()
```
