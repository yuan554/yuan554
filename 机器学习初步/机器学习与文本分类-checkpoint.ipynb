```python
import tensorflow as tf
from tensorflow import keras
from keras.datasets import imdb
import pandas as pd
import numpy as np
import nltk
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import TfidfVectorizer#文本特征抽取包 
from nltk.corpus import stopwords
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score
```

```python
import pandas as pd
iris=pd.read_csv("C:\\Users\\yuan\\Desktop\\lecture7\\酒店文本评价.csv",encoding="gbk")
#将数据集分为特征和目标变量
X=iris.label
y=iris.review
#将数据集分为训练集和测试集
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=0)
```

```python
X
```

```python
#创建随机森林分类器
rf=RandomForestClassifier(n_estimators=2,max_depth=5,random_state=9063)
#参数选择会影响模型性能，但没有一个单一的参数可以适用于所有场景。
#n_estimators：这个参数指定森林中决策树的数量,较大的值通常可以提供更好的性能，但也会增加计算时间和内存,数值通常在50-100
#max_depth：限制每棵决策树的最大深度，较大的值可以减少过拟合，但也会影响模型的预测能力，数值通常在10-20
#random_state：用于设置随机种子，以便结果可重复
```

```python
#训练模型
rf.fit(X_train,y_train)
```

```python
#预测测试集
y_pred_forest=rf.predict(X_test)
```

```python
#计算准确率
accuracy_forest=accuracy_score(y_test,y_pred_forest)
print(classification_report(y_test, y_pred_forest))
conf_mat_forest = confusion_matrix(y_test, y_pred_forest)
print(conf_mat_forest)
```

```python
#创建决策树分类器
tree_clf = DecisionTreeClassifier(random_state=42)
tree_clf.fit(X_train, y_train)
y_predtree = tree_clf.predict(X_test)
accuracy_tree = accuracy_score(y_test, y_predtree)
print(classification_report(y_test, y_predtree))
conf_mat_tree = confusion_matrix(y_test, y_predtree)
print(conf_mat_tree)
```

```python
import pandas as pd
data=pd.read_csv("C:\\Users\\yuan\\Desktop\\lecture7\\酒店文本评价.csv",encoding="gbk")
data
```

```python
data.sentiment.value_counts().plot(kind='pie')
```

```python
#划分输入变量和输出变量
x=data['sentimenttext']
y=data['sentiment']
#创建一个TfidfVectorizer的实例,min_df:词汇表中忽略文档频率低于该值的词,max_df:忽略文档频率高于该值的词
vec=TfidfVectorizer(max_features=2000,min_df=0.7,max_df=0.3)
```

```python
#使用Tfidf将文本转化为向量,U为unicode
x=vec.fit_transform(x.values.astype('U')).toarray()
x
```

```python
#划分训练集和测试集
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=0)
```

```python
from sklearn.ensemble import RandomForestClassifier
clf=RandomForestClassifier(n_estimators=250,random_state=0)
clf.fit(x_train,y_train)
```

```python
y_pred=clf.predict(x_test)
```

```python
#查看混淆矩阵
print(confusion_matrix(y_test,y_pred))
```

```python
#查看模型性能
print(classification_report(y_test,y_pred))
```

```python
print(accuracy_score(y_test,y_pred))
```
