```python
import nltk #导入英文自然语言处理工具包
```

```python
words=nltk.word_tokenize(txt)#英文分词
```

```python
words
```

```python
#词干提取
from nltk.stem.porter import PorterStemmer
porter_stemmer=PorterStemmer()
```

```python
#新建一个列表放入词干提取内容，以便后续逐一提取并判断其是否在停用词列表
lis=[]
for word in words:
    c=porter_stemmer.stem(word)
    lis.append(c)
```

```python
lis
```

```python
from nltk.corpus import stopwords
stopwords
```

```python
#查看自带的停用词列表
stop_words=stopwords.words("english")
stop_words
```

```python
#去除停用词
fitered_sentence=[w for w in lis if not w in stop_words]
fitered_sentence
```

```python
dist=nltk.FreqDist(fitered_sentence)#统计词频数
```

```python
dist.items
```

```python
for k in dist.items():
    print(k)
```

```python
#取出字典中的键值对
for k,v in dist.items():
    print(k,v)
```

```python
#查看出现频数最高的30个词
most=dist.most_common(30)
most
```

```python
#词云制作
wordcloud=WordCloud(background_color="white",
                    width=1999,height=888,margin=5).generate(str(fitered_sentence))
plt.imshow(wordcloud)
```

```python
c=stp.split("\n");c
```

```python
filtered_sentence=[w for w in words if not w in c];filtered_sentence
```

```python
#python字典的get方法会返回指定键的值,dict.get(‘键’),返回“键”对应的“值”,如果键不在字典中则返回默认值None
#过滤掉所有长度为1的词语
#合并同义词并进行词频统计
counts={}
for word in filtered_sentence:
    if len(word)==1:
        continue
    elif word=="大老":
        rword="天保"
    elif word=="二老":
        rword=="傩送"
    elif word=="爷爷":
        rword=="祖父"
    else:
        rword=word
    counts[rword]=counts.get(rword,0)+1
```

```python
counts.items()
```

```python
#将元素加入列表以便索引
items=list(counts.items())
items
```

```python
items[3]
```

```python
a=items.sort(key=lambda x:x[1],reverse=True)
a
```

```python
for i in range(len(items)):
    word,count=items[i]
    print(word,count)
```
