```python
source = open("C:\\Users\\yuan\\Desktop\\lecture 6\\热水器评价.csv","r",encoding="gbk")
line1 = source.readlines()
line1
```

```python
sentimentslist = []
for i in line1:
    s = SnowNLP(i)
    print(s.sentiments)
    sentimentslist.append(s.sentiments)
```

```python
# 设置中文字体
plt.rcParams['font.sans-serif'] = ['SimHei']  # 使用SimHei字体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题
```

```python
# 绘制柱状图
plt.figure(figsize=(10, 6))
plt.hist(sentimentslist, bins=np.arange(0, 1.01, 0.01), facecolor='y')  
plt.title('情感分数段出现频率')
plt.xlabel('情感分数段')
plt.ylabel('频率')
plt.xticks(rotation=45)
plt.show()
```

```python
# 将列表转换为DataFrame
df = pd.DataFrame(sentimentslist, columns=['sentiment'])

# 计算均值和中位数
mean_sentiment = df['sentiment'].mean()
median_sentiment = df['sentiment'].median()

# 打印结果
print(f"情感分数均值: {mean_sentiment:.2f}")
print(f"情感分数中位数: {median_sentiment:.2f}")
```
